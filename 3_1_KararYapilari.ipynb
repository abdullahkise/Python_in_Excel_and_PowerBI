```python
elif alternatifŞartTrueVeyaFalseUretir:
    #True ise bu bloga girer
elif alternatifŞart2:
    #True ise bu blok çalışır
else:
    #tüm şartların dışında bir sonuç dönerse (tüm şartlar False dönerse) bu bloga girer.
```

````python
#Soru:
# Dışarıdan girilen sayısal bir notun harf karşılığını yazdıralım.
# mesela 90 girilirse A yazsın. mesaj olarak notunuz 90 harf karşılığı A'dır desin.

#kullanıcıdan değer almak için input
sayiNotu= "60" #input("Notunuzu giriniz (0-100)")
sayiNotu = int(sayiNotu) #tip dönüşümü yaptık. str tipini int tipine çevirdik.

#
harfNotu = ""

#sayiNotu>=0 and sayiNotu<=100
if 0<=sayiNotu<=100:
    #if blogunun içidir. if şartından True gelirse burası çalışır
    #pass #bu keyword kodlar için yer tutucudur. Yani buraya kod yazacağım şimdilik geçiyorum hata verme demiş olduk.

    #0-100 arasında not girilmiş ise harf karşılıını bulalım.
    if sayiNotu>=85:
        harfNotu = "A"
    elif sayiNotu>=70:
        harfNotu = "B"
    elif sayiNotu>=50:
        harfNotu = "C"
    else:
        harfNotu = "D"

    #-------------
    print(f"Notunuzu {sayiNotu} olarak girdiniz. Notunuzun harf karşılığı {harfNotu} dır.")        
else:
    #şarttan false dönerse bu blog çalışır
    print("Notunuz 0-100 arasında olmalıdır. Hatalı giriş yaptınız.")
````

### match
* Match çok daha farklı amaçlar için kullanılabilir. Ancak switch yerine aşağıdaki şekilde kullanabiliriz
  
```python
match alternatifDegerlerAlacakDegisken:
    case alternatifDeger1:
        #alternatifDeger1 ile karşılaşınca yapılacaklar
    case alternatifDeger2:
        #alternatifDeger2 ile karşılaşınca yapılacaklar
    case alternatifDeger2:
        #alternatifDeger1 ile karşılaşınca yapılacaklar
    case _:
        #else blogu gibi çalışır.
```

````python
departman = "OlmayanBirDepartman"

match departman:
    case "Güvenlik":
        print("Görev yeriniz giriş katı")
    case "IT":
        print("Evden çalışabilirsiniz")
    case _: #else blogu gibi çalışıyor.
        print("IK ile görüşünüz")
````
