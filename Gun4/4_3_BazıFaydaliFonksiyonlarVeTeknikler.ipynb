```python
#kalan değeleri göz ardı etmek istersek. _ veya *_ kullanırız
x,_,y,*_,z= range(0,10)  #[0,2,3,4,5,6,7,8,9]

"""
    x: ilk deger
    _: bir deger atla
    y: üçüncü değer
    z: son değer
    *_: aradaki atla
"""

print(x,y,z, sep="\n")
```

### Zip, Map, Filter

#### Zip
* `zip(iter, iter)` iterasyon yapılbilecek nesnelerin elemanların aynı sırada getirirr. Paralel loop yapar. 

```python
#zip(iter, iter) #iterasyon yapılbilecek nesnelerin elemanların aynı sırada getirirr. Paralel loop yapar.

adlar = ["Ali","Veli","Ayşe"]
cinsiyetler = ["E","E","K"]

#
zip(adlar, cinsiyetler) #bir zip nesnesi döner.

#
list(zip(adlar, cinsiyetler)) #liste tipine çevirdiğimiz elemanları liste olarak aynı tuple içinde eşleşmiş şekilde görebiliriz.
```

```python
#for ile bu zip içinde dönmek mümkün
"""
for z in zip(adlar, cinsiyetler):
    print(z) 
"""

for z in zip(adlar, cinsiyetler): 
    #print(f"Adınız : {z[0]} ve Cinsiyetiniz: {z[1]}") #her bir eleman tuple olarak temsil. tuple sequence olduğu için index ile elemanlarına erişebiliriz.
    print("Adınız : {} ve Cinsiyetiniz: {}".format(*z)) #.format(z[0],z[1])) #*z ile z'nin elemanarını format fonksiyonu parametreleri olarak vermiş olduk
```

```python
for ad, cinsiyet in zip(adlar, cinsiyetler):
    print(f"adınız {ad} ve cinsiyetiniz {cinsiyet}") #for satırında gelen tuple o dongude unpack edip ad ve cinsiyet değişkenleri ile temsil
```

#### Map
* `map(fonksiyon, iter)` verilen fonksiyonu iter elemanlarına uygular. iter elemanlarını fonksiyona parametre olarak gönderir

```python
fiyatlar = [12.3, 334.6, 3544.1] 
#fiyatların kdv dahil hallerini hesaplalım

#
#fiyatlar + 5 #+ operatoru listlerde tanımlı değil. HATA!!!
#fiyatlar * 3 #bu tanımlı fakat beklenmedik şekilde eleman sayısını bu çarpan kadar arttırır.

#listelerde broadcast yani bir işlem istendiğinde tüm elemanlara uygulama özelliği mevcut değil.
#1. list comphrension
[fiyat*1.20 for fiyat in fiyatlar] #her bir eleman 1.20 ile çarpıldı geriye liste döndğü

#2. map ile 
hesaplama_map = map(lambda f:f*1.20 ,fiyatlar) 
#lambda f:f*1.20 fonksiyonunu tanımladık.
list(hesaplama_map)
```

#### filter
* `filter(fnk, iter)` fnk geriye True/false döndürmesi gerekir. True karşılğı olan elemanlar gelir diğerleri filtrelenir gelmez.

```python
lst = ["metin","abdullahkise@hotmail.com","metin2","****","123"]

#1. sadece alfabediki karekterlerden oluşanlar gelsin.
[l for l in lst if str.isalpha(l)] #l elemanı alfabedki harflerden mi oluşuyor.


#2.filter
list(
    filter(str.isalpha,lst) #() ile fonksiyon çalışır. Biz fonksiyonu işaret edeceğiz. o yüzden () kullanmıyoruz
)
```

## Regex

```python
import re
```

```python
metin = """
Merhaba

Herhangi bir soru olursa abdullahkise@hotmail.com adresinden benimle iletişime geçebilirsiniz. 
Eğer ulaşamazsanız abdullahkise@gmail.com adresimi kullanabilirsiniz.

Sağlıklı günler dilerim,

Kise
"""
```

```python
mail_expression = r'([\w\.-]+)@[\w\.-]+'
```

```python
#1.re.search #ilk bulduğunu getirir.
sonuc_match = re.search(mail_expression , metin) #match nesnesi döner. eşleşen yoksa None döner

#
sonuc_match.group(0) #ilk bulduğu eşleşen ifadeyi verir.
sonuc_match.group(1) #() ile belirtilen bölgeyi verir.
```

```python
#2.re.findall
re.findall(r"[\w\.-]+@[\w\.-]+" , metin)
```

```python
#3.re.split
csvMetin = "Ali Uçan      22            Müdür" #değişken ayraçlarda bir veri geldi.

#\s boşluk anlamına gelir. \s+ değişen uzunlukta boşluk
re.split(r"\s+" , csvMetin) #r ile raw string yani olduğu gibi metni kabul et. özel bir karakter arama içinde \n alt satıra geçmek gibi
```
