````python
#Cent. Fah. çeviren bir fonskiyon yazalım
def fnk_sicaklik_cevir(santigrad=15):
    fahrenayt = 1.8 * santigrad + 32
    return fahrenayt
````

````python
fnk_sicaklik_cevir(20)
fnk_sicaklik_cevir() #parametrenin varsayılan değeri kullanılır? parametrenin varsayıan değeri 15
````

````python
#* ile sınırsız sayıda değer kaubl edilebiliriz
#** ile sınırsız sayıda key-value çifti kabul eder
def fnk_sinirsiz(*sinirsizValue , **sinirsizKeyValue):
    print(sinirsizValue) #* ile tanımlanan parametreden gelen değerleri fonksiyon içinde tuple olarak tutulur.
    print(sinirsizKeyValue) #** ile tanımlanan parametreden gelen değerleri fonksiyon içinde dict tipinde temsil etti.

#-----------------------
#fnk_sinirsiz(1,2,3,4,5)
#
fnk_sinirsiz(1,2,3,4,5, ad="Ali", soyad="Uçan") #ad ve soyad key olarak alg
````

## Lambda yöntemi ile fonksiyon tanımlama
```python
lambda parametreler : returnEdilecekYaniGeriDondurulecek
```

````python
#1. klasik yöntem
def c2f(c):
    return c*1.8 + 32

#c2f(30)
#2.lambda
#lambda c : c*1.8 + 32 #bu bir fonksiyon nesnesi bunu () çalıştırabiliriz.

c2f = lambda c : c*1.8 + 32
c2f(30) #parantez açıp kapattığımızda değişkene atadığmız fonksiyonu çalıştırırız.----------------
````
