### for döngüsü
* iterasyon yapılabilebilecek (içinde nesneler olan ve bu nesnelere erişmeyle ilgili bir alt yapıya sahip olan nesneler) elemanlarına tek tek ulaşmamızı sağlar.
* for içerisinde bu ulaşılan elemanlarla çalışılabilir.
```python
for Eleman in ElemanContaineri:
    #her bir Eleman ile çalışabiliyoruz.
```

````python
kisiler = ["Ali","Veli","Ayşe"]

#
for kisi in kisiler:
    #her bir kisiler listesi elemına erişebileceğim. bu eleman içeride kisi değişkeni ile temsil edilecek.
    print(kisi.upper()) #str.upper büyük harfle yazmamızı sağlar.
````

#### range

````python
#sistem başka container tipler de var. Bunlar farklı kütüphanelerde veya standart kütüphaneler veya gömülü nesneler arasonda bulunabilir.
#en kullanışlı olanlardan biri de range fonksiyonudur.
#range(start , stop, step) # ile bir sayı grubu bize verir.
[0,1,2,3,4]

#range(0,5,1)
range(5) #geriye içinde 0,1,2,3,4 sayılarını olduğu bir nesne döner. Biz bu nesneyi list tipine çevirip vbakabiliriz. veya for ile range elemanlarına erişebiliriz.

#
list(range(5))
````

````python
#soru:
#1-10 arsındaki 3'e tam bölünen sayılara bakalım.
#sayilar = [1,2,3,4,5,6,7,8,9]
sayilar = range(1,11) #1-10 sayılarını tutar. for ile elemanlarına erişilebilir.

for s in sayilar:
    #her bir sayi s olarak temsil edilecek.
    #s 3'e bölümünden kalan sıfır ise yani tam bölünürse 
    if s%3==0:
        print(s)
````

````python
#soru:
#0-100 sayıları arasındaki 14'e tam bölünenlerin adedini, kendisini ve max, min değerlerini getirelim.
sayilar = range(101) #0-100 arasıdaki sayılar

#
lst_tamBolunenler = [] #boş liste olsun. tam bölünenleri burada toplayacağız.

#
for s in sayilar:
    if s%14==0:
        #print(s)
        lst_tamBolunenler.append(s)

# #for döngüsü bitti
# lst_tamBolunenler #tam bolunenlerin kendisi
# len(lst_tamBolunenler) #eleman sayısı
# max(lst_tamBolunenler) #en büyük değer
# min(lst_tamBolunenler) #en küçük değer

#
dict_sonuclar = {
    "tamBolunenler":lst_tamBolunenler,
    "adet":len(lst_tamBolunenler),
    "enBuyuk":max(lst_tamBolunenler),
    "enKucuk":min(lst_tamBolunenler)
}

#
dict_sonuclar
#---------------------------------
#mesajda bunları yazalım
## f-string yöntemi
# mesaj = f"""
# Verilen aralıktaki 14'e tam bölünen sayilar {dict_sonuclar["tamBolunenler"]} olarak tespit edildi.
# Bu sayıların adedi {dict_sonuclar["adet"]} dur.
# Sayıların en büyüğü {dict_sonuclar["enBuyuk"]} ve en küçüğü {dict_sonuclar["enKucuk"]} olarak bulundu.
# """

#format_map yöntemi
mesaj = """
Verilen aralıktaki 14'e tam bölünen sayilar {tamBolunenler} olarak tespit edildi.
Bu sayıların adedi {adet} dur.
Sayıların en büyüğü {enBuyuk} ve en küçüğü {enKucuk} olarak bulundu.
""".format_map(dict_sonuclar)

print(mesaj)
````

````python
#soru: 10-20 arsındaki sayılardan çift olanları getirelim.
#1. yöntem açık for ve geriye list döndürelim.
lst_ciftler = []
for r in range(10,21):
    if r%2==0:
        lst_ciftler.append(r)
#
lst_ciftler
````

````python
#2.list comphrension yöntemi
[r for r in range(10,21) if r%2==0] #sonuçta bir liste elde edeceğiz
````

### while-else
* while yanında belirtilen şart True ürettikçe döngü çalışır. ilk False ürettiğinde else blogu devreye girer. else blogu zorunlu değildir.
```python
while şart:
    #şart True oldukça bu blok çalışır
else:
    #şart'ın False ürettiği ilk noktada döngü durur. durmadan önce else bloguna uğrar.
```
* Tüm döngülerde `break` kelimesi görüldüğünde döngü durur. `continue` kelimesi görüldüğünde o noktadan aşağı inmez döngünün başına geçer sonrak adıma atlar

````python
sayac = 0

#
while sayac<10:
    #sayac = sayac + 1
    sayac += 1 #kümülatif atama operatoru ile aynı işi yapabiliriz.
    #---------------
    #not: if blokları bağımsızdır. yani her bir şart kontrol edilir bağımsız şekilde true üretirse gövdedeki kodları çalışır.
    if sayac == 7:
        break #burada artık döngüyü durdur. Döngü bu şekilde durusa else doğal olarak çalışmaz.

    if sayac %2==0:
        continue #burada dur ve iterasyona devam et.
    #---------------
    print(sayac)
else:
    print(f"Şarta uymayan bir durumla karşılaıldı {sayac =}") #place holder için degisken = kullanırsak değişken adı ile degerini birlikte yazar
````
