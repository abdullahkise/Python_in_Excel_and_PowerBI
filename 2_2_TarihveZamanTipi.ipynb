# Tarih ve Zaman Tipleri
* standart tipler arasında zaman ve tarih ile ilgili tipler de bulunmaktadır.
* Standart kütüphaneler arasında `datetime` modülü zaman ve tarih ile ilgili işlemler için kullanılmaktadır.
* `from datetime import datetime, date, time, timedelta`
  * datetime: tarih ve zaman ile ilgili işlemler için kullanılır.
  * date: sadece tarih ile ilgili işlemler için kullanılır.
  * time: sadece zaman ile ilgili işlemler için kullanılır.
  * timedelta: iki tarih/zaman arasındaki farkı temsil eder.

```python
from datetime import datetime, date, time, timedelta
```

```python
bugun = date.today()

#type(bugun) #datetime.date
#dir(bugun)

#
bugun.year
bugun.month
bugun.day

print(bugun) #print ile yazdırnda ISO formatında Yıl-Ay-Gun formatında yazar.
bugun #bu bir date nesnesi. Ekrana geliştiriciler ne göstermek istedi ise o görünüyor.
```

```python
simdi = datetime.today()

#
simdi.year
simdi.hour

#
simdi.date() #date nesnesi
simdi.date().year


#
simdi.time() #time nesnesi
simdi.time().hour
```

## Tarih Oluşturmak için

```python
#1. yöntem: tip üzerinden oluşturulabilir.
t = date(year=2026, month=9, day=22)
t #date tipinden tarih oluştu.
```

```python
#2. yöntem: metin üzerinden tarih oluşturulabilir.
#metnin hangi formatta girileceğini string format karakterleri ile belirtiriz.
#yıl için %Y, ay için %m ve gün için %d kullanılır. Diğerleri notlardaki resimde.
#strptime: zamanı parse etmek için kullanır.
t = datetime.strptime("22-09-2026", "%d-%m-%Y") #strptime sadece datetime tipi üzerinde mevcut

t #datetime
```

## Tarih Güncellemek

```python
t = t.replace(day = 23) #tarih gününü artık 23 yapıyoruz. orjinalini etkilelemez. tekrar yanı tipe atama yapalım

t
```

## Timedelta ile zaman farkı yönetimi

```python
zamanFarki = date(2026,12,31) - date.today() #yıl başına kadar geçen süre #datetime.timedelta(days=99) tipini veriyor.
#dir(zamanFarki)
zamanFarki.days
zamanFarki.total_seconds()
```

```python
# Bugünden 3 hafta sonrası hangi tarihe tekabül ediyor.
hafta = timedelta(days = 7) #7 günlük fark. - + olarak tanımlanabilir

bugun = date.today()

#
bugun + hafta*3  #zaman farkı matemtiksel işlemlerde kullanılabilir.(bazı)
```
