```python
#1. Mantık hatası
positifTamsayi = 5

if positifTamsayi<0:
    print("Bu mesaj hiç görünmez. Çünkü mantıken her zaman pozitif sayı vereceğimiz değişkeni negatif mi diye kontrol ediyoruz.")
```

```python
#2. Syntax hatası
x = 12

#= atama operatoru == ise karşılaştırma operatorodur.
if x%2=0:
    print("çift Sayıdır")
```

```python
#3. run-time hataları
x = 12
y = 0 #"metin" #0

#
try:
    sonuc = x/y

    #
    #print(sonuc)
except ZeroDivisionError as ex:
    print("Bölüm sıfır olamaz.")
    print("Sistem hata mesajı : ",ex)
except TypeError:
    print("Veri tipi hatası var. Veriler sayısal olmalıdır.")
#except:  #hata olması durumda çalışacak kodlar 
# Exception tipi en genel hatayı temsil eder. tüm hataları için tetiklenir.
except Exception as ex:
    print(f"Bir hata ile karşılaşıldı. Sistem hata mesajı : {ex}")
else:
    #hata olmadğında devreye girer
    print("Program hata olmadan tamamlandı.")
    print("Sonuc: ",sonuc)
finally:
    #hata olsa olmasa da devreye girer. Dosya ve veritbanı bağlantıları kapatılabilir.
    print("Programın sonuna gelindi.")

    
```
