```python
# Soru: Bir çalışan hakedişini hesaplayalım.
"""
   - minimum 40 saat çalışma olacağını kabul edelim. Kullanıcı bunun üzerinde çalışma girsin.
   - 40 saatin üzerindeki çalışmalar normal hakedişin 1.5 katı olsun.
   - saatlik hakediş değiştirilebilir olsun.
"""
calismaSaati  = 67
#
saatlikHakedis = 554.7

#hakedişi hesaplalım
hakedis = 40 * saatlikHakedis + (saatlikHakedis*1.5)*(calismaSaati - 40)

#
hakedis

#
round(hakedis, 2) #virgülden sonra 2 basamak gösterdi. Bu hala bir sayı.
```

## String Formatlama
1. str.format
2. f-string
3. str.format_map(dict)

```python
#{} placeholder yer tutucudur. Buraya değişkenler gelen değerler sonra eklenecek. "".format(,,,,) ile placeholderlar doldurulur
#{deger : formatstring}
#s tring, f float, d int kullanılır
#1. Yöntem str.format
print("{} saat çalıştınız. Hakedişiniz {} TL".format(calismaSaati,hakedis))
print("{} saat çalıştınız. Hakedişiniz {:.3f} TL".format(calismaSaati,hakedis)) #float hane 3 basamaklı oldu.
print("{} saat çalıştınız. Hakedişiniz {:.1f} TL".format(calismaSaati,hakedis)) #float hane 1 basamak oldu.
print("{} saat çalıştınız. Hakedişiniz {:_.1f} TL".format(calismaSaati,hakedis)) #, veya _ ile binlik ayraç verilebilir.
print("{} saat çalıştınız. Hakedişiniz {:>30_.1f} TL".format(calismaSaati,hakedis)) #<,^,> sol, orta, sağ hizalar.
print("{} saat çalıştınız. Hakedişiniz {:*>30_.1f} TL".format(calismaSaati,hakedis)) #kalan boşluklara * işareti koyduk.

#2.Yöntem f-string
print(f"{calismaSaati} saat çalıştınız. Hakedişiniz {hakedis:*>30_.1f} TL")
```

## Kullanıcıdan Değer Alma ve Tip Dönüşümü
* Kullanıcıdan değer almak için input() fonksiyonu kullanılır. Kullanıcın girdiği değerleri tipi str'dir
  *  eğer sayısal bir işleme tabi tutmak istersek tip dönüşmü yapmak gerekir

```python
s1 = input("1. Sayıyı giriniz: ")
s2 = input("2. Sayıyı giriniz:")

toplam = s1+s2

print(toplam)
```
