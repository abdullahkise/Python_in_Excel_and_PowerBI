## Bazı Array Fonksiyonları

```python
dizi = np.arange(1,25).reshape(4,6) #reshape ile yeni bir shape (boyut verebiliriz.)

dizi
```

```python
dizi.sum() #dizinin tüm elemanlarını toplar.

#istersek bunlara yön yani axis vereibliriz.
#axis=0 satır boyunca işlem yap. satırları topla
#axis=1 sutunlar üzerinde işlem yap. Sutunları topla

dizi.sum(axis=0)
dizi.sum(axis=1)

#min, max, mean, std, cumsum vs. gibi fonksiyonlar mevcut.
```

```python
#satırlar ile sutunları yer değiştirir.
dizi.transpose()
dizi.T
```

## Numpy Elemanlarına Erişmek
* Subsetting
* Slicing
* Boolen Indexing

```python
dizi
```

### Subsetting

```python
dizi[0] #iki boyutlu bir dizidir. İlk boyutunun yanin satırın 0. sını getir demiş olduk.
dizi[2]

#
dizi[2][3] #ilk boyutun (satır oluyor) 2. elemanını ve ikinci boyurun (kolon) 3. elemanını getir.

#tek seferde yapalım
dizi[2,3] #satırların 2. ve sutunların 3.sunu getir
```

### Slicing
`dizi[start:stop:step , start:stop:step,....]` şeklinde her boyut için sliver verilebilir

```python
"""
    satırda 2:5:3 ile
        - 2. indexten 5. indexe kadar 3'er adımdakini getir
    sutunda 1:4:2
        - 1. indexten 4. indexe kadar 2'şer adımdakini getir
"""
dizi[2:5:3,1:4:2]
```

```python
#negatif indexler kullanılabilir.
dizi[:2 , :-1] #satırlardan baştan 2. satıra, sutunlardan baştan başla sondan bir önceki kadar al.

dizi[:, ::-1] #step -1. yani tüm satırlar gelsin. kolonları tersten getirsin.
```

### Boolen Indexing
* True/False dizisi filtre olarak kullanılır

```python
#broadcast: her bir elemana talep edlen işlem uygulanır.
dizi < 5 #her bir eleman 5'ten küçük mü? True ve False dönecek.
```

```python
#örn:
#3'e tam bölünenleri bulalım
# ve 2'ye tam bölünenler
# her ikisine de aynı anda bölünenler.
filtre_3 = dizi %3 ==0
filtre_2 = dizi %2 ==0

#
filtre = filtre_2 & filtre_3 #np.logical_and(filtre_2,filtre_3)
#logical_and, logical_or, logical_and fonksiyonları and, or, not gibi çalışır.
#& ile np.logical_and
#| ile np.logical_or
#! ise sağdaki dizi için np.logical_not yerine geçer.

#
dizi[filtre] #filtre dizi, dizi ile aynı eleman sayısına ve boyutuna sahip, eleman eleman bakılır. True denk gelen eleman seçilmiş olur.
```
