```python
#örn.
# içerisinde ar geçen kolonları ve ad kolonu getirelim. Tüm satırlarla birlikte gelsin.
filtre_ad = ekip.columns == "Ad" #broadcast özelliği olan bir veri tipi
filtre_ar = ekip.columns.str.contains("ar") #column.str broadcast özelliğini destekleyen bir alt module

#
filtre = filtre_ad | filtre_ar #np.logical_or  # array([False,  True, False,  True,  True])
#
ekip.loc[:, filtre ]
```

```python
#soru:
'''
    1. Sadece Ad, Cinsiyet, Uretim kolonlarını çekelim.
    2. Uretim 150'den büyük satırları getirelim.
    3. 300'den küçük üretime sahip olmayan Erkek üyeler gelsin.
    4. 150'den büyük üretime sahip olan kadın üyeler.
'''
#1.
ekip.loc[:, ["Ad","Cinsiyet","Miktar"]]

#2.
filtre_miktar = ekip.Miktar > 150
ekip.loc[filtre_miktar]

#3.
filtre_erkek = ekip.Cinsiyet =="E"
filtre_300denKucuk = ekip.Miktar < 300

filtre = filtre_erkek & (~filtre_300denKucuk) #AltGr + Ü ile ~ işarei koyduk. np.logical_not anlamına gelir. Her bir elemanın tersini verir.

ekip.loc[filtre, ["Ad","Cinsiyet"]]

#4. soru sizde.
```

### `NULL (None, NaN, NaT, N/A)` değerlerine erişmek
* `isna()` veya `isnull()` ile null mı diye kontrol edilebilir.
* `notna()` veya `notnull()` ile null değil değil mi?
* `fillna()` ile null değerleri doldurabilirizi
* `dropna()` null değerleri kaldırabiliriz

```python
ekip.isna() #tüm df'te null var mı hücre hücre True/False
ekip.Miktar.isna() #eleman eleman null mı diye kontrol etti.
```

```python
#sum ile Truları toplayabilir any ile True var mı diye kontrol edebiliriz.
ekip.isna().any() #kolon kolon null varsa True döndü
ekip.isna().any().any() #bir önceki any'den dönen True/False serisinde True var mı?
```

```python
ekip.isna().sum() #kolon kolın nulların adedi
ekip.isna().sum().sum() #toplamda kaç tane null var görebiliriz.
```

```python
#örn. Miktar alanı null olan alanları getirelm. bunlara ortalama değeri basalım.
ekip.Miktar.isna() #bunu filtre olarak kullanalım

#kolon karşısına = ile değer ataması yapabilirsiniz. eğer bu isimde bir kolon yoka o kolon oluşur. hata vermez.
ekip.loc[ekip.Miktar.isna(), "Miktar"] = ekip.Miktar.mean()

ekip
```

```python
#null değerlere otomatik değer atamak
ekip.fillna("Bilinmiyor") #ortak bir değer vermek için

#hangi kolonda null varsa ne yazsın
ekip.fillna({
                "Miktar":ekip.Miktar.mean(),
                "Cinsiyet":"Bilinmiyor",
                "DepartmanId":-1
})
```

```python
#dropna ile null alanlar kaldırılabilir
ekip.dropna() #içinde bir şekilde null geliyorsa o satırları kaldırır.

#
ekip.dropna(axis=1, how="any") #axis=1 kolon anlamına gelir. #how="any" varsayılan davranıştır. herhangi bir null varsa kolonu kaldırır.
ekip.dropna(axis=1, how="all") #tüm değerleri null ise o kolonu kaldırır

#axis=0 ile satır kaldırılabilir.
```
