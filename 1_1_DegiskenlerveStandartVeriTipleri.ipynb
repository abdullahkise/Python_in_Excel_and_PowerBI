### Container Elemanlarına Erişim

#### Sequence (Sıralı ve indekslenebilir container tipleri) (list, tuple, str)
* index ile elemanlara erişim
  * index numarası 0'dan başlar
  * negatif index ile sondan elemanlara erişim
  * `nesne[index]`
* slicer ile erişim
  * start:end:step formatı kullanılır
  * start dahil, end hariçtir
  * step ile atlamalı erişim sağlanabilir
  * `nesne[start:stop:step]`
    * start belirtilmezse en baştan başlanır
    * stop belirtilmezse sona kadar gidilir
    * step belirtilmezse varsayılan olarak 1 alınır

```python
#index yöntemi ile elemanlara erişim
metin = "Merhaba"

metin[0] #0.indexteki yani ilk elemana erişiriz.
metin[3] #3.indexteki elemanı getrir.

#
metin[-1] #son elemana erişiriz.
```

```python
#slicer ile
lst = [10,20,30,"Python","Excel","PowerBI"]

#
#lst[0] # ilk elemana erişim

lst[0:3] #ilk 3 elemana erişim 1'er atlayarak [10, 20, 30]
lst[1:4] #2.,3. ve 4. elemana erişim [20, 30, 'Python']

#
lst[:3] #ilk 3 elemana erişim [10, 20, 30] start belirtilmediği için en baştan getirdi
lst[2:] #stop belirtilmediği için sona kadar getirdi [30, 'Python', 'Excel', 'PowerBI'] 2. indexten sona kadar

#
lst[::2] #2'şer atlayarak tüm elemanlara erişim [10, 30, 'Excel']

#step -1 ile elemanları tersten getirir.
lst[::-1] #tüm elemanlara ters sırayla erişim ['PowerBI', 'Excel', 'Python', 30, 20, 10]
```

#### Mapping Tipe Erişmek
